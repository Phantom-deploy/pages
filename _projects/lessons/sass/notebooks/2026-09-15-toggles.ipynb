## Switches (Refactor Toggles for open-coding)


### What Are Toggles?
Toggles are a kind of switch. You use them to turn things on and off, or check criteria.

You probably use toggles every day without even realizing it! Toggles are used in apps, websites, and devices to quickly turn settings on or off.

---

### What Are Toggles Good For?

Toggles let users quickly turn a setting on or off. They are commonly used in apps and websites for things like:

- Notifications
- Dark Mode
- Privacy settings
- Battery settings
- Website features

Toggles work best when there are two choices, such as on/off or enabled/disabled.

A toggle can use a selection statement to decide what happens based on whether the checkbox is checked.

For example, the logic for a toggle can be represented in College Board pseudocode:

IF the checkbox is checked
    Change the slider color
    Move the slider button to the right
ELSE
    Keep the slider gray
    Keep the slider button on the left

The IF/ELSE statement represents selection because the toggle behaves differently depending on whether the checkbox is checked.

In the UI Runner below, this idea is implemented using the `:checked` selector.

### UI Runner



````python
%%html
<!-- UI_RUNNER: Toggle example | show_code: true -->
<!-- //Fill in the blanks! -->
 
<style>
.switch {
    position: relative;
    display: flex;
    width: 54px;
    height: 31px;
}

.switch input[type="checkbox"] {
    opacity: 0;
    width: 0;
    height: 0;
}

.slider {
    position: absolute;
    inset: 0;
    border: 1px solid var(--ui-border, #666);
    border-radius: 100px;
    background: var(--panel-mid, gray);
    cursor: pointer;
}

.slider:before {
    content: "";
    position: absolute;
    width: 23px;
    height: 23px;
    top: 3px;
    left: 4px;
    bottom: 0;
    border-radius: 50%;
    background: var(--pref-text-color, white);
    transition: transform 0.2s;
}

input[type="checkbox"]:checked + .slider {
    background: var(--pref-accent-color, blue);
}

input[type="checkbox"]:checked + .slider:before {
    transform: translateX(22px);
    outline: 2px solid var(--pref-accent-color, blue);
}
</style>

<h3>Checkbox</h3>
<label>
  <input type="checkbox">
</label>

<h3>Slider</h3>
<label class="switch">
  <input type="checkbox">
  <span class="slider" aria-hidden="true"></span>
  <span>Toggle</span>
</label>
````

### Code Breakdown

1. Set Up the `.switch`
   - Set the toggle's position, size, padding, and alignment.

2. Set Up the Checkbox
   - Select the checkbox and make it invisible while keeping it functional.

3. Style the `.slider`
   - Create the visible toggle with its position, rounded shape, background color, and transition.

4. Create the Slider Button
   - Use `&:before` to create the white circle and set its size, position, and shadow.

5. Add the Checked State
   - Use `&:checked + .slider` to change the toggle's color when it is turned on.
   - The `.green` class can change the checked color to `$green1`.

6. Move the Slider Button
   - Use `translateX(27px)` to move the white circle when the toggle is turned on.

7. Add Focus and Transitions
   - Add a focus shadow with `&:focus + .slider`.
   - Use `transition` to make the toggle changes smooth.

````python
%%html
<!-- UI_RUNNER: Toggle example | show_code: true -->
<!-- //Fill in the blanks! -->
 
<div>
  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle1" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you have family members who have or have had cancer?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle2" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you consume alcohol or use tobacco?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle3" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you forget to wear sunscreen?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle3" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you consume less than 48 oz of water and 30 grams of fiber a day?</span>
  </label>

  <br>

  <button id="checkRisk">Check Risk</button>

  <p id="riskOutput"></p>
</div>

<script>
  let risk = 0;

  document.getElementById("checkRisk").addEventListener("click", function() {
    
    // Start risk at 0 each time the button is pressed
    risk = 0;

    // Find all toggles
    const toggles = document.querySelectorAll(".ocs__toggle-input");

    // Add 25 for every checked toggle
    toggles.forEach(function(toggle) {
      if (toggle.checked) {
        risk += 25;
      }
    });

    // Print the risk
    document.getElementById("riskOutput").textContent = "Current Cancer Exposure Risk: " + risk + "%";
  });
</script>
````

## The Beauty of SASS

To use checkboxes or sliders all you need to do is use the class from SASS!  This lets you focus on logic instead of repeating style over and over.

````python
%%html
<!-- UI_RUNNER: Toggle example using SASS | show_code: true -->

<label class="ocs__toggle ocs__toggle--green">
  <input id="feature-toggle" class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">My Label</span>
</label>

Status: <output id="feature-toggle-status" aria-live="polite"></output>

<script>
{
  const toggle = document.getElementById("feature-toggle");
  const status = document.getElementById("feature-toggle-status");

  const updateStatus = () => {
    status.value = String(toggle.checked);
  };

  toggle.addEventListener("change", updateStatus);
  updateStatus();
}
</script>
````
